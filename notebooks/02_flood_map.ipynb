## Reading the result honestly
- `s1_unobservable_fraction` and `s1_valid_fraction` in the report above are the share of the AOI the radar could not judge (layover/shadow, no data). Flood area is a lower bound there.
- Debris bit 0 = Sentinel-1 evidence, bit 1 = Sentinel-2 vegetation loss. Neither is validated against ground truth.
- Thresholds are defaults (or tuned on Kuro Siwo *train* patches via `make tune`), never tuned on Trishuli.
